```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
```

# Stage 1

```python
df1 = pd.read_parquet("ltws_oiwfs_data_16.parquet")
df1.head()
```

## WFE RMS & Segment Piston

```python
fig,axs = plt.subplots(nrows=2,sharex=True,figsize=(15,10))
ax = axs[0]
ax.plot(np.vstack(df1["SegmentWfeRms<-9>"]))
ax.legend(list(range(1,8)))
ax.plot(np.vstack(df1["WfeRms<-9>"]),'k')
ax.grid()
ax.set_ylabel("WFE RMS [nm]")
ax = axs[1]
ax.plot(np.vstack(df1["SegmentPiston<-9>"]))
ax.grid()
ax.set_xlabel("Steps")
ax.set_ylabel("Segment Piston [nm]")
```

## First & last wavefronts

```python
w = np.vstack(df1["Wavefront b"])
fig,axs = plt.subplots(ncols=2, figsize = (16,6))
ax = axs[0]
h = ax.imshow(w[0].reshape(512,-1))
fig.colorbar(h,ax=ax)
ax = axs[-1]
h = ax.imshow(w[-1].reshape(512,-1))
fig.colorbar(h,ax=ax)
```

# Stage 2

```python
df2 = pd.read_parquet("ltws_oiwfs_dfs_data_1.parquet")
df2.head()
```

## WFE RMS & Segment Piston

```python
fig,axs = plt.subplots(nrows=2,sharex=True,figsize=(15,10))
ax = axs[0]
ax.plot(np.vstack(df2["SegmentWfeRms<-9>"]))
ax.plot(np.vstack(df2["WfeRms<-9>"]),'k')
#ax.set_xlim([0,2000])
ax.grid()
ax.set_ylabel("WFE RMS [nm]")
ax = axs[1]
ax.plot(np.vstack(df2["SegmentPiston<-9>"]))
ax.grid()
#ax.set_xlim([0,2000])
ax.set_xlabel("Steps")
ax.set_ylabel("Segment Piston [nm]")
```

```python
fig,axs = plt.subplots(nrows=2,sharex=True,figsize=(15,10))
ax = axs[0]
ax.plot(np.vstack([np.vstack(df1["SegmentWfeRms<-9>"]),np.vstack(df2["SegmentWfeRms<-9>"])]))
ax.plot(np.vstack([np.vstack(df1["WfeRms<-9>"]),np.vstack(df2["WfeRms<-9>"])]),'.')
ax.grid()
ax.set_ylabel("WFE RMS [nm]")
ax.set_xlim([0,100])
ax = axs[1]
ax.plot(np.vstack([np.vstack(df1["SegmentPiston<-9>"]),np.vstack(df2["SegmentPiston<-9>"])]))
ax.grid()
ax.set_xlim([0,100])
ax.set_xlabel("Steps")
ax.set_ylabel("Segment Piston [nm]")
```

## M1 rigid body motions

```python
fig,ax = plt.subplots()
ax.plot(np.vstack(df2["M1RigidBodyMotions"]));
ax.grid()
ax.set_xlabel("Steps")
```

## M1 bending modes

```python
fig,ax = plt.subplots()
ax.plot(np.vstack(df2["M1ModeShapes"]));
ax.grid()
```

## M2 asm command

```python
fig,ax = plt.subplots()
ax.plot(np.vstack(df2["M2ASMAsmCommand"]));
ax.grid()
```
